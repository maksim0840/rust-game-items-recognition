## Рассчитать парамтры сетки

Для построения модели сетки и определения координат ячеек достаточно 3 значения: начальная координата сетки, размер ячейки, размер пустого пространства между ячейками.

Зададим модель сетки следующими числами:
- x0 - начальная координата x
- y0 - начальная координата y
- cell_w - длинна ячейки
- cell_h - высота ячейки
- delta_w - (ширина ячейки + зазор до следующей ячейки в ряду) ~ шаг
- delta_h - (высота ячейки + зазор до следующей ячейки в колонке) ~ шаг

Определим данные параметры по координатам крайних точек ячеек и уравнениям прямых:
- x_начало_ячейки = x0 + col · delta_w
- y_начало_ячейки = y0 + row · delta_h

```python
import numpy as np

# раскладывает замеры на начала и концы
def _starts_ends(coords, comp):
    """coords: список пар (x, y). comp=0 -> x (ось X), comp=1 -> y (ось Y).
       Значения идут start, end, start, end, ... -> чётные=начала, нечётные=концы."""
    vals = [p[comp] for p in coords]
    return np.array(vals[0::2], float), np.array(vals[1::2], float)

# превращает замеры одной оси в 3 параметра (нулевая координата, шаг до следующей ячейки, размер ячейки)
def _fit_axis(starts, ends):
    """origin и delta — совместным фитом прямой (не усреднением!);
       cell_size — усреднением по ячейкам."""
    j = np.arange(len(starts))
    delta, origin = np.polyfit(j, starts, 1)   # наклон = delta (шаг), свободный член = origin
    cell = float(np.mean(ends - starts))
    return float(origin), float(delta), cell


chest_params = []

# собираем информацию о ящике с новыми параметрами
for cfg in chest_interface:
    sx, ex = _starts_ends(cfg['row_cords'], 0)   # X — из верхнего ряда
    sy, ey = _starts_ends(cfg['col_cords'], 1)   # Y — из левого столбца
    x0, delta_w, cell_w = _fit_axis(sx, ex)
    y0, delta_h, cell_h = _fit_axis(sy, ey)
    chest_params.append({
        'width':  cfg['width'],
        'height': cfg['height'],
        'scale':  cfg['scale'],
        'x0':      x0,
        'y0':      y0,
        'cell_w':  cell_w,
        'cell_h':  cell_h,
        'delta_w': delta_w,
        'delta_h': delta_h,
    })

display(chest_params)
```

## Определение уравнений для коэффициентов сетки

Построим графики зависимости для новых расчитанных коэффициентов сетки: x0, y0, cell_w, cell_h, delta_w, delta_h

По графикам сырых значений в зависимости от scale видно, что зависимость - линейная. То есть для описания данных коэффициентов подойдёт линейное уравнение.

Отбросив разрешение экрана из значений коэффициентов, видно, что точки на графике совпали. То есть коэффициенты сетки одинаковые на одной и той же доле экрана (scale), если отбросить влияние разрашения экрана. Значит, разрешение экрана - это множитель, который их различает (коэффициенты сетки пропорциональны разрешению экрана).

## Уравнения точек интерфейса

- x = WIDTH · (a · SCALE + k)
- y = HEIGHT · (a · SCALE + k)

или:
- x = a · WIDTH · SCALE + k · WIDTH
- y = a · HEIGHT · SCALE + k · HEIGHT


## Построение уравнений для предсказания параметров сетки

- x0 =          a1 · (WIDTH · SCALE) + k1 · WIDTH
- delta_w =     a2 · (WIDTH · SCALE) + k2 · WIDTH
- cell_w =      a3 · (WIDTH · SCALE) + k3 · WIDTH
- y0 =          a4 · (HEIGHT · SCALE) + k4 · HEIGHT
- delta_h =     a5 · (HEIGHT · SCALE) + k5 · HEIGHT
- cell_h =      a6 · (HEIGHT · SCALE) + k6 · HEIGHT

```python
from sklearn.linear_model import LinearRegression

# points — chest_params (список конфигураций с параметрами x0,y0,cell_w,cell_h,delta_w,delta_h)
# param  — какой параметр предсказываем: 'x0','delta_w','cell_w','y0','delta_h','cell_h'
# axis   — по какому размеру он масштабируется: 'x' -> width, 'y' -> height
def fit_param(points, param, axis):
    dim_key = 'width' if axis == 'x' else 'height'
    X, y = [], []
    for p in points:
        X.append([p[dim_key] * p['scale'], p[dim_key]])   # [X1 = dim*scale, X2 = dim]
        y.append(p[param])
    reg = LinearRegression(fit_intercept=False).fit(X, y)
    a, k = reg.coef_
    return a, k
```

```python
grid_model_args = {
    'x0':       fit_param(chest_params, 'x0',       'x'),
    'delta_w':  fit_param(chest_params, 'delta_w',  'x'),
    'cell_w':   fit_param(chest_params, 'cell_w',   'x'),
    'y0':       fit_param(chest_params, 'y0',       'y'),
    'delta_h':  fit_param(chest_params, 'delta_h',  'y'),
    'cell_h':   fit_param(chest_params, 'cell_w',   'y'),
}

# Параметры уравнений a,k для предсказания коэффициентов по известным width, height, scale
for name, (a, k) in grid_model_args.items():
    print(f"{name:8s}: a={a:.8f}  k={k:.8f}")
```

## Детектор ячеек

```python
from PIL import Image

class CellsDetector:
    # Посчитанные параметры для определения коэффициентов сетки
    GRID_MODEL_ARGS = {
        'x0':      (0.15538758, 0.49977814),    # * width
        'delta_w': (0.04845760, -0.00002766),   # * width
        'cell_w':  (0.04531963, -0.00045448),   # * width
        'y0':      (-0.83747928, 0.99939329),   # * height
        'delta_h': (0.08609323, 0.00002244),    # * height
        'cell_h':  (0.08054795, -0.00079300),   # * height
    }

    # Предсказывает коэффициенты сетки по известному разрешению экрана и scale
    @staticmethod
    def predict_grid_params(width, height, scale):
        def v(param, dim):
            a, k = CellsDetector.GRID_MODEL_ARGS[param]
            return dim * (a * scale + k)
        return {
            'x0':      v('x0', width),
            'delta_w': v('delta_w', width),
            'cell_w':  v('cell_w', width),
            'y0':      v('y0', height),
            'delta_h': v('delta_h', height),
            'cell_h':  v('cell_h', height),
        }

    # Возращаем координаты всех ячеек сундука
    # Каждая ячейка: (tl_x, tl_y, br_x, br_y) tl — верхний левый угол, br — нижний правый.
    @staticmethod
    def all_slot_boxes(width, height, scale, n_cols, n_rows):
        p = CellsDetector.predict_grid_params(width, height, scale)
        x0, dw, cw = p['x0'], p['delta_w'], p['cell_w']
        y0, dh, ch = p['y0'], p['delta_h'], p['cell_h']
        boxes = []
        for row in range(n_rows):
            for col in range(n_cols):
                boxes.append((
                    round(x0 + col*dw),        # x1
                    round(y0 + row*dh),        # y1
                    round(x0 + col*dw + cw),   # x2
                    round(y0 + row*dh + ch),   # y2
                ))
        return boxes

    # Нарезает изображение сундука на ячейки по модели сетки
    @staticmethod
    def extract_cells(image, scale, n_cols, n_rows, inset=3):
        """
        Возвращает список из n_cols*n_rows словарей:
        {'row', 'col', 'box' (рамка без отступа), 'image' (кроп PIL.Image)}
        """
        # 1) приводим вход к PIL.Image
        if isinstance(image, str):
            img = Image.open(image).convert('RGB')
        elif isinstance(image, np.ndarray):
            img = Image.fromarray(image)
        else:
            img = image
        W, H = img.size

        # 2) координаты всех ячеек по модели
        boxes = CellsDetector.all_slot_boxes(W, H, scale, n_cols, n_rows)

        # 3) режем
        cells = []
        for idx, (x1, y1, x2, y2) in enumerate(boxes):
            row, col = divmod(idx, n_cols)
            cx1 = max(0, min(W, x1 + inset)); cy1 = max(0, min(H, y1 + inset))
            cx2 = max(0, min(W, x2 - inset)); cy2 = max(0, min(H, y2 - inset))
            crop = img.crop((cx1, cy1, cx2, cy2))
            cells.append({'row': row, 'col': col, 'box': (x1, y1, x2, y2), 'image': crop})
        return cells
```

```python
# пример получения координат ячеек
cells = CellsDetector.all_slot_boxes(2560, 1440, 0.80, n_cols=6, n_rows=8)
for cell in cells:
    print(cell)
```

```python
def show_cells_grid(cells, n_cols, n_rows, figscale=1.4):
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * figscale, n_rows * figscale))
    for cell in cells:
        ax = axes[cell['row']][cell['col']]
        ax.imshow(cell['image'])
        ax.set_title(f"r{cell['row']}c{cell['col']}", fontsize=6)
        ax.axis('off')
    plt.tight_layout()
    plt.show()

# пример получения изображений ячеек
cells = CellsDetector.extract_cells("/content/Снимок экрана 2026-07-27 201555.png", scale=1, n_cols=6, n_rows=8) # 2560x1440
show_cells_grid(cells, n_cols=6, n_rows=8)
```
